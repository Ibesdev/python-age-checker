```python
import tkinter as tk

age = int(input("Enter your age: "))


fenster = tk.Tk()          
fenster.title("Age Checker")
fenster.geometry("400x300")
fenster.configure(background="#414141") # bg (backround) ist Hintergrund 
#des Fensters man kann name oder hexcode verwenden


anzeige = tk.Label(fenster, font=("Arial", 20), fg="black") # fg= schriftfarbe
anzeige.pack(pady=50)

if age >= 100:
    anzeige.config(text="You are dead bro.")
elif age >= 18:
    anzeige.config(text="You are eligible to vote.")
else:
    anzeige.config(text="You can't vote yet.")



#Fürs fenster das es verschiedene antworten geben kann muss
#anzeige.config(text="")

fenster.mainloop()

```
